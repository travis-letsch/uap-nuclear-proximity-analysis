```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException
from selenium.common.exceptions import NoSuchElementException
from webdriver_manager.chrome import ChromeDriverManager
import time
import random
```

<br>

## Overview
This is the first of three notebooks for this project. This notebook will scrape the 2023, 2024, and 2025 data from the NUFORC database. The CSV files will be used in the second notebook, tletsch_2_data_prep.ipynb to clean the data and prepare it for analysis.

<br>

#### 2023 Scraping Loop

```python
# Initialize the webdriver
browser = webdriver.Chrome()
browser.maximize_window()
```

```python
# Create empty lists to store the scraped data
occurred = []
city = []
state = []
country = []
shape = []
summary = []
```

```python
# Loop to scrape the NUFORC website
wait = WebDriverWait(browser, 10)

# Range 1-13 to capture all 12 months of data
for month_begin in range(1, 13):
    month_str = f"{month_begin:02d}"

    # Updaute 202(3) in the URL to switch years
    url = f"https://nuforc.org/subndx/?id=e2023{month_str}"

    print(f"Scraping: {url}")

    browser.get(url)
    time.sleep(random.uniform(2, 4))

    # Page counter to track pages for each month of data
    page_num = 1

    while True:
        print(f"  Scraping page {page_num}")
        
        # Wait for table rows to load before scraping
        wait.until(
            EC.presence_of_element_located((By.XPATH, "//table[@id='table_1']/tbody/tr"))
        )

        rows = browser.find_elements(By.XPATH, "//table[@id='table_1']/tbody/tr")

        for row in rows:
            try:
                occurred_date = row.find_element(By.CSS_SELECTOR, "td.column-occurred").text
                sighting_city = row.find_element(By.CSS_SELECTOR, "td.column-city").text
                sighting_state = row.find_element(By.CSS_SELECTOR, "td.column-state").text
                sighting_country = row.find_element(By.CSS_SELECTOR, "td.column-country").text
                ufo_shape = row.find_element(By.CSS_SELECTOR, "td.column-shape").text
                sight_summary = row.find_element(By.CSS_SELECTOR, "td.column-summary").text

                occurred.append(occurred_date)
                city.append(sighting_city)
                state.append(sighting_state)
                country.append(sighting_country)
                shape.append(ufo_shape)
                summary.append(sight_summary)

            except Exception:
                print(f"{month_str}, page {page_num}: row skipped")

        # Find the Next button
        next_button = browser.find_element(By.ID, "table_1_next")

        # Stop if Next button is disabled
        if "disabled" in next_button.get_attribute("class"):
            print(f"  Finished {month_str}")
            break

        # Click next page
        browser.execute_script("arguments[0].click();", next_button)
        
        # The page data loads irregularly. Add a pause to wait for tables to refresh before scraping
        wait.until(EC.staleness_of(rows[0]))  # wait for old rows to disappear

        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "#table_1 tbody tr")))

        page_num += 1
```

<br>

```python
# Populate a DataFrame with the scraped data
nuforc_2023 = pd.DataFrame({
    "occurred": occurred,
    "city": city,
    "state": state,
    "country": country,
    "shape": shape,
    "summary": summary
})
```

```python
# Display the DataFrame and export it to a CSV file
display(nuforc_2023)

nuforc_2023.to_csv("nuforc_2023.csv", index=False)
```
